# Cyclist data analysis
----------------------------------

```python
import pandas as pd
```

```python
q1_2019 = pd.read_csv("Divvy_trips_data/Divvy_Trips_2019_Q1.csv")
q1_2020 = pd.read_csv("Divvy_trips_data/Divvy_Trips_2020_Q1.csv")
print("Data loaded successfully!")
```

#============================================
#### Wrangling & Combining Data into a Single File
#============================================

```python
# Checking the columns for each df
print("Columns for Q1 2019:\n", q1_2019.columns)
print("\nColumns for Q1 2020:\n", q1_2020.columns)
```

```python
# Renaming the columns in q1_2019 to match the column names in q1_2020 for consistency
q1_2019 = q1_2019.rename(columns={
    'trip_id':'ride_id',
    'bikeid':'rideable_type',
    'start_time':'started_at',
    'end_time':'ended_at',
    'from_station_name':'start_station_name',
    'from_station_id':'start_station_id',
    'to_station_name':'end_station_name',
    'to_station_id':'end_station_id',
    'usertype':'member_casual'
})

# Inspecting the data frames to make sure everything correct.
print("\nQ1 2019 Data info:")
q1_2019.info()
print("\nQ1 2020 Data info:")
q1_2020.info()
```

```python
# Converting the ride_id and rideable_type columns to string type for consistency
q1_2019["ride_id"] = q1_2019["ride_id"].astype(str)
q1_2019["rideable_type"] = q1_2019["rideable_type"].astype(str)
```

```python
# Stacking the two dataframes together to create a single dataframe for analysis
all_trips = pd.concat([q1_2019, q1_2020], ignore_index=True)
```

```python
# Dropping unnecessary columns that are not needed for the analysis
all_trips = all_trips.drop(columns=['start_lat', 'start_lng', 'end_lat', 'end_lng', 'birthyear', 'gender', 'tripduration'], errors='ignore')
```

#===========================================
### Cleaning and Adding Data for Analysis
#===========================================

```python
# inspecting the combined dataframe to make sure everything is correct.
print("\nList of columns after combining & dropping unnecessary columns:\n", all_trips.columns)
print("\nHow many rows are in the combined dataframe:\n", len(all_trips))
print("\nDimensions of the dataframe (rows, columns):\n", all_trips.shape)
print("\nfirst 6 rows of the combined dataframe:\n", all_trips.head())
print("\nList of columns and data types:\n")
all_trips.info()

#
print("\nStatistical summary of the dataframe:\n", all_trips.describe(include='all'))
```

```python
# Checking how many observations are in each category of the member_casual column
print("\nValue counts for the member_casual column:\n", all_trips["member_casual"].value_counts())
```

```python
# replacing the values in the member_casual column to make them more readable and easier to understand
all_trips["member_casual"] = all_trips["member_casual"].replace({"Subscriber": "member", "Customer": "casual"})
```

```python
# checking to make sure the proper number of observations are in each category of the member_casual column after the replacement
print("\nValue counts for the member_casual column:\n", all_trips["member_casual"].value_counts())
```

```python
# Converting the 'started_at' and 'ended_at' columns to datetime format
all_trips["started_at"] = pd.to_datetime(all_trips["started_at"])

all_trips["ended_at"] = pd.to_datetime(all_trips["ended_at"])
```

```python
all_trips.info()
```

```python
# add columns that list the date, month, day, year, and day of week of each ride
all_trips["date"] = all_trips["started_at"].dt.date
all_trips["month"] = all_trips["started_at"].dt.month
all_trips["day"] = all_trips["started_at"].dt.day
all_trips["year"] = all_trips["started_at"].dt.year
all_trips["day_of_week"] = all_trips["started_at"].dt.day_name()
```

```python
# Calculate the length of each ride in seconds
all_trips["ride_length"] = (all_trips["ended_at"] - all_trips["started_at"]).dt.total_seconds()

# Display information about the DataFrame including the new columns
print("\nData types after adding new columns:\n")
all_trips.info()
```

```python
# checking for trips with ride lengths negative or zero
all_trips[all_trips["ride_length"] <= 0]
```

```python
#* checking for trips with ride lengths greater than 24 hours (1440 minutes)
all_trips[all_trips["ride_length"] > 1440]
```

```python
# Removing the rows with negative ride lengths and rides that start at "HQ QR" by filtering the dataframe and creating a new dataframe called all_trips_v2 using copy() to avoid the SettingWithCopyWarning. This will ensure that the original dataframe remains unchanged and we can work with a clean dataset for analysis.
all_trips_v2 = all_trips[(all_trips["start_station_name"] != "HQ QR") & (all_trips["ride_length"].between(0, 1440))].copy()
```

#============================
### DESCRIPTIVE ANALYSIS
#============================

```python
# Descriptive analysis of the ride_length column to understand the distribution of ride lengths in the dataset to identify any potential outliers or anomalies in the data that may need to be addressed before further analysis.
print("\nDescriptive statistics for ride_length column:\n", all_trips_v2["ride_length"].describe())
```

The descriptive statistics for ride_length show clear anomalies: the maximum value exceeds 10 million seconds (~123 days) and the standard deviation is far larger than the mean. 

These patterns indicate the presence of extreme outliers or erroneous ride duration records. 

```python
# grouping the data by member_casual and calculating the mean, median, max, and min ride lengths for each group to understand how the ride lengths differ between members and casual riders.
print("\nRide Length statistics grouped by member_casual:\n", all_trips_v2.groupby('member_casual')['ride_length'].agg(['mean','median','max','min']))
```

```python
# Average ride length by member type and day of the week
print("\nAverage ride length by member type and day of the week:\n")
all_trips_v2.groupby(['member_casual','day_of_week'])['ride_length'].mean()
```

Across both groupby analyses (ride length statistics by member_casual, and average ride length by member_casual and day_of_week), the results show a clear data‑quality issue in the ride_length column for casual riders.

Member ride durations fall within realistic ranges (mean around 13 minutes, median around 8 minutes). In contrast, casual riders show extremely inflated values, including a maximum ride length of 10,632,022 seconds (~123 days) and mean values exceeding 5,300 seconds (~1.5 hours). These durations are not plausible for bike‑share usage and strongly indicate the presence of corrupted or erroneous records (e.g., trips not properly ended, timestamp errors, or multi‑day rides).

These outliers significantly distort the mean and max values for casual riders in both groupby results. Before drawing conclusions or comparing member vs casual behavior, the dataset should be cleaned by removing or correcting unrealistic ride durations.

```python
# Average ride length by member type and day of the week
print("\nAverage ride length by member type and day of the week:\n")
avg_ride_length = all_trips_v2.groupby(['member_casual','day_of_week'])['ride_length'].mean()

avg_ride_length
```

```python
# Ordering the days of the week to ensure that the analysis is done in a logical order from Monday to Sunday. This will help in visualizing and interpreting the data more effectively.
days_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
all_trips_v2["day_of_week"] = pd.Categorical(all_trips_v2["day_of_week"], categories=days_order, ordered=True)

```

```python
# Running again the aggregation to see the average ride length by member type and day of the week after ordering the days of the week.
print("\nAverage ride length by member type and day of the week:\n", all_trips_v2.groupby(['member_casual','day_of_week'], observed=True)['ride_length'].mean())
```

```python
# Calculating summary statistics by member type and day of the week, including the number of rides and average duration of rides for each group. This will provide insights into the riding patterns of members and casual riders throughout the week.
summary_stats = all_trips_v2.groupby(['member_casual','day_of_week'], observed=True).agg(
    number_of_rides = ('ride_id', 'count'),
    average_duration = ('ride_length', 'mean')
    ).reset_index()


print("\nSummary statistics by member type and day of the week:\n", summary_stats)
```

```python
# Saving the summary statistics to a CSV file for further analysis or reporting.
counts = all_trips_v2.groupby(['member_casual',
'day_of_week'], observed=True)['ride_length'].mean().reset_index()
counts.to_csv('avg_ride_length_v3.csv', index=False)


counts
```
